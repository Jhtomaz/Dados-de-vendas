```python
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\Colaborador IP\Documents\GitHub\Dados-de-vendas\data\Base_Vendas_Aula_PowerBI - Vendas.csv', decimal=',')
print(df.head())
```

```python
def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    print("Normalizando nomes das colunas...")
    df_sales = df.rename(
        columns={
            "Data":"data",
            "ID_Venda":"id_venda",
            "ID_Cliente":"id_cliente",
            "ID_Produto":"id_produto",
            "ID_Vendedor":"id_vendedor",
            "Quantidade":"quantidade",
            "Preco_Unitario":"preco_unitario",
        }
    )

    print("Nomes das colunas normalizados com sucesso.")
    return df_sales
```

```python
normalized_df = normalize_columns(df)
print(normalized_df.head())
```

```python
print(normalized_df.dtypes)
```

```python
def column_types(df: pd.DataFrame) -> pd.DataFrame:
    print("Alterando tipos de dados das colunas...")
    df = df.astype({
        'data': 'datetime64[ns]',
        'id_venda': 'string',
        'id_cliente': 'int64',
        'id_produto': 'int64',
        'id_vendedor': 'int64',
        'quantidade': 'int64',
        'preco_unitario': 'float64'
    })
    print("Tipos de dados das colunas alterados com sucesso.")
    return df
```

```python
column_types_df = column_types(normalized_df)
print(column_types_df.dtypes)
```
